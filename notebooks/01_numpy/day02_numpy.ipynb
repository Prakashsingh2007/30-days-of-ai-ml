```python
import numpy as np

# Sample array containing numbers above and below 250
arr = np.array([120, 255, 80, 310, 245, 400, 190])

# Filter and print numbers greater than 250
greater_than_250 = arr[arr > 250]

print(greater_than_250)
# Output: [255 310 400]
n5 = greater_than_250.mean()
print("Mean of numbers greater than 250:", n5)

n6 = arr[[0,4,5]]
print("Selected elements at indices 0, 4, and 5:", n6)
```
