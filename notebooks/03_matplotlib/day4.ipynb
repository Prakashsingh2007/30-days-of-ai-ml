```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x, y)

plt.show()
```

```python
plt.plot(x,y)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.plot(
    x,
    y,
    marker="o"
)

plt.title("Study Hours vs Marks")

plt.show()
```

```python
branches = ["CSE","AI","ECE"]

students = [120,90,60]

plt.bar(branches, students)

plt.show()
```

```python
import numpy as np

marks = np.random.randint(
    0,
    100,
    500
)

plt.hist(marks)

plt.show()
```

```python
labels = [
    "Python",
    "Java",
    "C++"
]

sizes = [
    60,
    25,
    15
]

plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%"
)

plt.show()
```

```python
hours = [1,2,3,4,5]

math = [45,60,70,80,90]

science = [50,55,68,75,95]

plt.plot(
    hours,
    math,
    label="Math"
)

plt.plot(
    hours,
    science,
    label="Science"
)

plt.legend()

plt.show()
```
