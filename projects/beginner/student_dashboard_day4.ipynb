```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

students = [f"Student {i}" for i in range(1, 101)]

study_hours = np.random.randint(1, 11, 100)   
marks = np.random.randint(35, 101, 100)           
attendance = np.random.randint(60, 101, 100)      
branch = np.random.choice(
    ["CSE", "IT", "ECE", "ME", "CE"],
    size=100
)
```

```python
df = pd.DataFrame({
    "Student": students,
    "Study Hours": study_hours,
    "Marks": marks,
    "Attendance": attendance,
    "Branch": branch
})

print(df.head())
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["Student"], df["Marks"], label="Marks", color="blue")
plt.title("Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.xticks(rotation=90)
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.savefig("line_chart.png")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df["Study Hours"], df["Marks"],
            color="red",
            label="Students")

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.legend()

plt.savefig("scatter_plot.png")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.hist(df["Marks"],
         bins=10,
         color="green",
         edgecolor="black",
         label="Marks Distribution")

plt.title("Histogram of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(True)
plt.legend()

plt.savefig("histogram_marks.png")
plt.show()
```

```python
branch_counts = df["Branch"].value_counts()

plt.figure(figsize=(7,5))
plt.bar(branch_counts.index,
        branch_counts.values,
        color="orange",
        label="Students")

plt.title("Students per Branch")
plt.xlabel("Branch")
plt.ylabel("Number of Students")
plt.grid(axis="y")
plt.legend()

plt.savefig("bar_chart_branch.png")
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(branch_counts.values,
        labels=branch_counts.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Branch Distribution")
plt.legend(branch_counts.index)

plt.savefig("pie_chart_branch.png")
plt.show()
```
