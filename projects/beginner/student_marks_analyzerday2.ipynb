```python
import numpy as np
n1 = np.random.randint(1,100, size=10)
print("Random integers between 1 and 100:", n1)
```

```python
print(np.max(n1))
print(np.min(n1))
print(np.mean(n1))
print(n1[n1 >= 40])
print(n1[n1 > 90])
```

```python
n2 = n1 + 5
print(n2)
print(n2[np.clip(n2, 0, 100) < 100])
print(np.sort(n2))
print(n2[-3:])
```
