```python
import numpy as np
import pandas as pd
a = np.random.randint(1, 100, 100)
print(a)
```

```python
print("Mean:", np.mean(a))
print("Median:", np.median(a))
print("Standard Deviation:", np.std(a))
print("Variance:", np.var(a))
print("Minimum:", np.min(a))
print("Maximum:", np.max(a))
```

```python
att = np.random.randint(a)
print(att)
```

```python
import pandas as pd
df = pd.DataFrame({
    "Attendance": att,
    "Marks": a
})
print(df)
```

```python
rel = df.corr()
print(rel)
avg = df[['Marks']].mean()
print(avg)
avg_above = df[df['Marks'] > avg['Marks']]
print(avg_above[['Marks']])
```

```python
att_above = df[df['Attendance'] > 75]
print(att_above[['Attendance']])
```

```python
df.to_csv('student_statistics_day3.1.csv', index=False)
```
